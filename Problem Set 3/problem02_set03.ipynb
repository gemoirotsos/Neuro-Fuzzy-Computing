- make three arrays to check when to present p1,p2 or p3

```python
import numpy as np

size = 1000
#select p1 array
select_p1_array = np.arange(0,size,3)

#select p2 array
select_p2_array = np.arange(1,size,3)


#select p3 array
select_p3_array = np.arange(2,size,3)
```

# compet activation function

```python
#compet activation fucntion
def compet(w,p,b):
    a = w@p + b.reshape(-1,1)

    #find max from a
    max_element = np.max(a)

    if max_element == a[0]:
        print("w1 wins")
        metric=1                #metric=1 -> w1 wins
    elif max_element == a[1]:
        print("w2 wins")
        metric=2                #metric=2 -> w2 wins
    elif max_element == a[2]:
        print("w3 wins")
        metric=3                #metric=3 -> w3 wins

    return metric       #metric is used so that i can know who wins




```

# kohonen rule

```python
#kohonen rule

def kohonen(w_old, p,learning_rate):
    w_new = w_old +  learning_rate * (p - w_old)
    return w_new
```

#  calculate the weights (using the Kohonen rule) and the biases (using the above bias rule). Repeat the sequence shown below until neuron 1 wins the competition 

```python
#input vectors
p1 = np.array([[-1],[0]])
p2 = np.array([[0],[1]])
p3 = np.array([[1/np.sqrt(2)],[1/np.sqrt(2)]])

#weights
w1 = np.array([[0],[-1]])
w2 = np.array([[-2/np.sqrt(5)],[-1/np.sqrt(5)]])
w3 = np.array([[-1/np.sqrt(5)],[-2/np.sqrt(5)]])

# w array
w = np.array([w1.flatten(), w2.flatten(), w3.flatten()])

#biases
b1=0
b2=0
b3=0
b = np.array([b1,b2,b3])   

#data
max_iter = 100
learning_rate=1

#sequence p1,p2,p3,p1,p2,p3,....

for i in range(max_iter):

    #weight array
    w = np.array([w1.flatten(), w2.flatten(), w3.flatten()])  
    #biases array
    b = np.array([b1,b2,b3])
    
    #present p1
    if i in select_p1_array:

        print(f"present p1 - iteration i={i}")


        a = compet(w,p1,b)                          #compet activ function

        if a==1:                                    #w1 wins
            print()
            print(f"##### w1 wins in iteration {i} #######")
            break
        elif a==2:                                  #w2 wins
            w2_new = kohonen(w2,p1,learning_rate)   #kohonen rule for w2
            w2=w2_new
            #biases
            b2 -= 0.2                               
            b3 *= 0.9                                  
        elif a==3:                                  #w3 wins
            w3_new = kohonen(w3,p1,learning_rate)   #kohonen rule for w3
            w3=w3_new
            #biases
            b2 *= 0.9                               
            b3 -= 0.2 
        print()

    #present p2
    elif i in select_p2_array:

        print(f"present p2 - iteration i={i}")

        a = compet(w,p2,b)                          #compet activ function

        if a==1:                                    #w1 wins
            print()
            print(f"##### w1 wins in iteration {i} #######")
            break
        elif a==2:                                  #w2 wins
            w2_new = kohonen(w2,p2,learning_rate)   #kohonen rule for w2
            w2=w2_new
            #biases
            b2 -= 0.2                               
            b3 *= 0.9
        elif a==3:                                  #w3 wins
            w3_new = kohonen(w3,p2,learning_rate)   #kohonen rule for w3
            w3=w3_new
            #biases
            b2 *= 0.9                               
            b3 -= 0.2 
        print()

    #present p3
    elif i in select_p3_array:

        print(f"present p3 - iteration i={i}")

        a = compet(w,p3,b)                          #compet activ function

        if a==1:                                    #w1 wins
            print()
            print(f"##### w1 wins in iteration {i} #######")
            break
        elif a==2:                                  #w2 wins
            w2_new = kohonen(w2,p2,learning_rate)   #kohonen rule for w2
            w2=w2_new
            #biases
            b2 -= 0.2                               
            b3 *= 0.9
        elif a==3:                                  #w3 wins
            w3_new = kohonen(w3,p2,learning_rate)   #kohonen rule for w3
            w3=w3_new
            #biases
            b2 *= 0.9                               
            b3 -= 0.2 
        print()
           
```
