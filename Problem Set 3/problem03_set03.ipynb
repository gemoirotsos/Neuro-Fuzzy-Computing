```python
from math import exp
import numpy as np
import matplotlib.pyplot as plt
```

```python
#tanh function
def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x)) 

#activation function
def NFC(x):
    return x*tanh(np.exp(x))

X = np.linspace(-10,10,100)
y = NFC(X)
y_2 = tanh(X)

#plot
plt.figure(figsize=(10,6))
plt.plot(X,y,label='Activation Function')
plt.plot(X,y_2,label='Tanh', color='red')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```
