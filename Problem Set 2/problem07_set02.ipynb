```python
import numpy as np

# Define the input image (5x5)
image = np.array([[5, 5, 2, 5, 5],
                  [5, 5, 2, 5, 5],
                  [7, 7, 5, 7, 7],
                  [5, 5, 2, 5, 5],
                  [5, 5, 2, 5, 5]])

# Define the filter (3x3)
filter_ = np.array([[1, 0, -1],
                    [2, 0, -2],
                    [1, 0, -1]])

# Get dimensions of the image and filter
image_height, image_width = image.shape
filter_height, filter_width = filter_.shape

# Calculate output dimensions (valid padding)
output_height = image_height - filter_height + 1
output_width = image_width - filter_width + 1

# Initialize the output feature map
output = np.zeros((output_height, output_width))

# Perform convolution (cross-correlation)
for i in range(output_height):
    for j in range(output_width):
        # Extract the region of interest
        region = image[i:i + filter_height, j:j + filter_width]
        # Perform element-wise multiplication and sum the result
        output[i, j] = np.sum(region * filter_)

# Print the output feature map
print("Output Feature Map:")
print(output)
```
