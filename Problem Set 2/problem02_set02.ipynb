# Problem 02 - Set 02

- Find the minimum of the function f(w) = w1^2 + w2^2* 10^6 using the Newton method and fixed step λ_k=1


```python
import numpy as np

def f(w):
    return w[0]**2 + 10**6 * w[1]**2

def gradient(w):
    return np.array([2*w[0], 2 * 10**6 * w[1]])

def hessian(w):
    return np.array([[2,0], [0, 2 * 10**6]])
    
def newton(f, grad_f, hessian_f, w0, tol, max_iter=100):
    w = np.array(w0)
    for i in range(max_iter):
        g_k = grad_f(w)
        H_k = hessian_f(w)

        if np.linalg.norm(g_k) < tol:
            print(f"g_k={g_k}")
            break

        w = w - H_k.T*g_k

        return w

#initial guess
w_0 = [1,1]
tol = 1e-6
max_iter=100

min = newton(f,gradient,hessian,w_0,tol,max_iter=100)
print(min)
det = np.linalg.det(min)
print(det)
if det>0:
    print("Minimum value of f:", f(min))
elif det<0:
    print("Maximum value of f:", f(min))



        
        


```
