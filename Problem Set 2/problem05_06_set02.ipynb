# Problem 05 - set 02

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt 

# Define symbols
w1, w2 = sp.symbols('w1 w2')

# Define the function f(w1, w2)
f = w1**2 + w2**2 * 10**6  # Example: f(w1, w2) = w1^2 + w2^3
print(f)

# Compute the gradient (partial derivatives)
gradient_f = [sp.diff(f, w1), sp.diff(f, w2)]

# Display the gradient
print("Gradient:", gradient_f)

w1_value = np.random.uniform(-40,-30)
w2_value = np.random.uniform(-40,-30)
w1_start = w1_value     #we will use in the next question so that initial values will be the same
w2_start = w2_value     #we will use in the next question so that initial values will be the same
print(f"w1={w1_value}, w2={w2_value}")    #initial guess

#data for AdaGrad
max_iter = 1000
G=np.zeros(2)
a=0.2      #learning rate
e = 1e-8    #small constant
tol = 1e-6  #tollerance
trajectory = [(w1_value, w2_value)]

for i in range(max_iter):
    g = np.array([float(grad.subs({w1: w1_value, w2: w2_value}).evalf()) for grad in gradient_f])
    G[0] += g[0] ** 2
    G[1] += g[1] ** 2
    w1_value -= a * g[0] / (np.sqrt(G[0]+e))
    w2_value -= a * g[1] / (np.sqrt(G[1]+e))

    trajectory.append((w1_value, w2_value))

    if np.linalg.norm(g) < tol:
        break

print(f"minimization for {g}")  

# Convert trajectory to NumPy array for plotting
trajectory = np.array(trajectory)

# Create a grid for contour plot
w1_vals = np.linspace(-40, 40, 100)
w2_vals = np.linspace(-40, 40, 100)
W1, W2 = np.meshgrid(w1_vals, w2_vals)
F = W1**2 + 1000000 * W2**2  # Replace with your function


# Plot the contour and trajectory
plt.figure(figsize=(10, 8))
plt.contour(W1, W2, F, levels=np.logspace(0, 7, 35), cmap='jet')
plt.plot(trajectory[:, 0], trajectory[:, 1], 'r-o', markersize=4, label="Optimization Path")
plt.colorbar(label='f(w1, w2)')
plt.xlabel('w1')
plt.ylabel('w2')
plt.title('Trajectory of Optimization on f(w1, w2)')
plt.legend()
plt.show()
```
