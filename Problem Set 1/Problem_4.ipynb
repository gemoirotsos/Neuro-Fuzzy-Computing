```python
import numpy as np
import matplotlib.pyplot as plt

# Ορίζουμε τη συνάρτηση log-sigmoid
def logsig(x):
    return 1 / (1 + np.exp(-x)) 

# Αρχικοποιούμε τις παραμέτρους σύμφωνα με την εικόνα
w11_1 = -3
w12_1 = -1.5
b1_1 = -1
b1_2 = -0.25

w21_1 = 1
w21_2 = 1.5
b2 = 2

# Δημιουργούμε ένα εύρος τιμών για το p από -3 έως 3
p_values = np.linspace(-3, 3, 100)

# Λίστες για την αποθήκευση των εξόδων
a1_1_values = []
a1_2_values = []
a2_values = []

# Υπολογίζουμε τις εξόδους για κάθε τιμή του p
for p in p_values:
    # Log-sigmoid layer
    n1_1 = w11_1 * p + b1_1
    a1_1 = logsig(n1_1)
    
    n1_2 = w12_1 * p + b1_2
    a1_2 = logsig(n1_2)
    
    # Αποθήκευση των εξόδων του log-sigmoid layer
    a1_1_values.append(a1_1)
    a1_2_values.append(a1_2)
    
    # Linear layer
    n2 = w21_1 * a1_1 + w21_2 * a1_2 + b2
    a2 = n2  # purelin
    a2_values.append(a2)

# Δημιουργία διαγραμμάτων
plt.figure(figsize=(12, 8))

# Διάγραμμα για a1_1
plt.subplot(3, 1, 1)
plt.plot(p_values, a1_1_values, label=r'$a_1^1$', color='blue')
plt.xlabel('p')
plt.ylabel(r'$a_1^1$')
plt.title('Log-Sigmoid Output (Neuron 1)')
plt.grid(True)

# Διάγραμμα για a1_2
plt.subplot(3, 1, 2)
plt.plot(p_values, a1_2_values, label=r'$a_1^2$', color='green')
plt.xlabel('p')
plt.ylabel(r'$a_1^2$')
plt.title('Log-Sigmoid Output (Neuron 2)')
plt.grid(True)

# Διάγραμμα για a2
plt.subplot(3, 1, 3)
plt.plot(p_values, a2_values, label=r'$a^2$', color='red')
plt.xlabel('p')
plt.ylabel(r'$a^2$')
plt.title('Linear Layer Output')
plt.grid(True)

plt.tight_layout()
plt.show()
```
