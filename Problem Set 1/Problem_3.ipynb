```python
import matplotlib.pyplot as plt
import numpy as np

# my data
c = 3.5
x0_a = 0.5
x0_b = 0.51

#create an array to store the values
N = 21
array_a = np.zeros(N)
array_b = np.zeros(N)

array_a[0] = x0_a
array_a[1] = c * x0_a * (1-x0_a)

array_b[0] = x0_b
array_b[1] = c * x0_b * (1-x0_b)


#generate x1 to x20
i=1
for i in range(N-1):
    array_a[i+1] = c * array_a[i] * (1 - array_a[i])

    array_b[i+1] = c * array_b[i] * (1 - array_b[i])

#print the values of x1 to x20 for x0=0.5
print(f"x0={x0_a}")
print("-----------------------")
i=0
for i in range(N):
    print(f"x[{i}] = {array_a[i]}")

#print the values of x1 to x20 for x0=0.51
print()
print(f"x0={x0_b}")
print("-----------------------")
for i in range(N):
    print(f"x[{i}] = {array_b[i]}")

#plot the trajectory for x0=0.5
plt.figure(figsize=(10,6))
plt.plot(range(N), array_a,marker='o', linestyle='-', label="Trajectory")
plt.title(f"Trajectory Plotting for x0={x0_a}")
plt.xlabel("Iteration (n)")
plt.ylabel("x[n]")
plt.xticks(range(N+1))
plt.grid(True)
plt.legend()
plt.show()

#plot the trajectory for x0=0.51
plt.figure(figsize=(10,6))
plt.plot(range(N), array_b, marker='*', linestyle='-', color='red', label="Trajectory")
plt.title(f"Trajectory Plotting for x0={x0_b}")
plt.xlabel("Iteration (n)")
plt.ylabel("x[n]")
plt.xticks(range(N+1))
plt.grid(True)
plt.legend()
plt.show()

#same plot for x0=0.5 and x0=0.51
plt.figure(figsize=(10,6))
plt.plot(range(N), array_a,marker='o', linestyle='-', color='blue', label="Trajectory")
plt.plot(range(N), array_b, marker='*', linestyle='-', color='red', label="Trajectory")
plt.title(f"Trajectory Plotting for x0={x0_b} (red) and x0={x0_a} (blue)")
plt.xlabel("Iteration (n)")
plt.ylabel("x[n]")
plt.xticks(range(N+1))
plt.grid(True)
plt.legend()
plt.show()

```

Watching the first two plots (for x0=0.5 and x0=0.51) we end up to the following conclusions:
1. The two trajectories are periodic. That's because the values of x[n] are pretty mych repeated. Also, the behavior of our system is repeated.
2. A difference in 0.01 for x0_a and x0_b doesn't affect the overal shape of the discrete trajectory. The red plot has overlap the blue one, which means that the two plots are pretty much same.
