```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# Ορισμός της συνάρτησης f(x)
def f(x, A, B, C, D, E):
    return np.log(1 + np.exp(A * (x + B) + C * x**2)) - np.log(1 + np.exp(D * (x - B))) + E

# Ορισμός της συνάρτησης ReLU
def relu(x):
    return np.maximum(0, x)

# Συνάρτηση απώλειας (Mean Squared Error)
def loss(params, x):
    A, B, C, D, E = params
    y_pred = f(x, A, B, C, D, E)
    y_true = relu(x)
    return np.mean((y_pred - y_true) ** 2)

# Δημιουργία τιμών για το x
x = np.linspace(-10, 10, 400)

# Αρχικές τιμές για τις παραμέτρους
initial_params = [1, 0, 0, 0, 0]

# Ελαχιστοποίηση της συνάρτησης απώλειας
result = minimize(loss, initial_params, args=(x,), method='BFGS')

# Λήψη των βέλτιστων παραμέτρων
A_opt, B_opt, C_opt, D_opt, E_opt = result.x
print(f'Optimal parameters: A={A_opt:.4f}, B={B_opt:.4f}, C={C_opt:.4f}, D={D_opt:.4f}, E={E_opt:.4f}')

# Υπολογισμός της συνάρτησης f(x) με τις βέλτιστες παραμέτρους
y_approx = f(x, A_opt, B_opt, C_opt, D_opt, E_opt)
y_relu = relu(x)

# Σχεδιασμός των συναρτήσεων
plt.figure(figsize=(10, 6))
plt.plot(x, y_approx, label='Approximated Function', linestyle='--')
plt.plot(x, y_relu, label='ReLU Function', linestyle='-')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Approximation of ReLU with Optimized Parameters')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# Definition of the function f(x)
def f(x, A, B, C, D, E):
    return np.log(1 + np.exp(A * (x + B) + C * x**2)) - np.log(1 + np.exp(D * (x - B))) + E

# Definition of the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Loss function (Mean Squared Error)
def loss(params, x):
    A, B, C, D, E = params
    y_pred = f(x, A, B, C, D, E)
    y_true = sigmoid(x)
    return np.mean((y_pred - y_true) ** 2)

# Generate x values
x = np.linspace(-10, 10, 400)

# Initial parameter values
initial_params = [1, 0, 0, 0, 0]

# Minimize the loss function
result = minimize(loss, initial_params, args=(x,), method='BFGS')

# Get the optimal parameters
A_opt, B_opt, C_opt, D_opt, E_opt = result.x
print(f'Optimal parameters: A={A_opt:.4f}, B={B_opt:.4f}, C={C_opt:.4f}, D={D_opt:.4f}, E={E_opt:.4f}')

# Compute the function f(x) with the optimal parameters
y_approx = f(x, A_opt, B_opt, C_opt, D_opt, E_opt)
y_sigmoid = sigmoid(x)

# Plot the functions
plt.figure(figsize=(10, 6))
plt.plot(x, y_approx, label='Approximated Function', linestyle='--')
plt.plot(x, y_sigmoid, label='Sigmoid Function', linestyle='-')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Approximation of Sigmoid with Optimized Parameters')
plt.legend()
plt.grid(True)
plt.show()
```
