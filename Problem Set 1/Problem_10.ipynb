```python
import numpy as np
import matplotlib.pyplot as plt

# Sigmoid function
def logsig(x):
    return 1 / (1 + np.exp(-x))

def logsig_derivative(x):
    return logsig(x) * (1 - logsig(x))

# Linear activation
def purelin(x):
    return x

def purelin_derivative(x):
    return 1

# Initialize parameters
p = 1  
gt = 1 + np.sin((np.pi / 3) * p)  

# Learning rates to test
learning_rates = [0.04,0.08,0.1,0.3]

# Initial weights and biases
W1_init = np.random.uniform(-0.5, 0.5, (10, 1))
b1_init = np.random.uniform(-0.5, 0.5, (10, 1))
W2_init = np.random.uniform(-0.5, 0.5, (1, 10))
b2_init = np.random.uniform(-0.5, 0.5, (1, 1))

# Storing the results for plotting
results = []

for a in learning_rates:
    # Reset weights and biases for each learning rate
    W1 = W1_init.copy()
    b1 = b1_init.copy()
    W2 = W2_init.copy()
    b2 = b2_init.copy()

    # Forward pass
    n1 = np.dot(W1, [[p]]) + b1  
    a1 = logsig(n1)  

    n2 = np.dot(W2, a1) + b2  
    a2 = purelin(n2)  

    # Compute error
    e = gt - a2  

    # Backpropagation

    # Sensitivity of second layer
    s2 = -2 * purelin_derivative(n2) * e  

    # Sensitivity of first layer
    F1_prime = np.diagflat(logsig_derivative(n1))  
    s1 = np.dot(F1_prime, np.dot(W2.T, s2))  

    # Update weights and biases
    W2_new = W2 - a * np.dot(s2, a1.T)  
    b2_new = b2 - a * s2  

    W1_new = W1 - a * np.dot(s1, np.array([[p]]).T)  
    b1_new = b1 - a * s1  

    # Store the updated weights and biases for this learning rate
    results.append({
        'learning_rate': a,
        'W1_new': W1_new,
        'b1_new': b1_new,
        'W2_new': W2_new,
        'b2_new': b2_new,
        'error': e[0][0]
    })

# Print results
for result in results:
    print(f"Learning Rate: {result['learning_rate']}")
    print("Updated W1:\n", result['W1_new'])
    print("Updated b1:\n", result['b1_new'])
    print("Updated W2:\n", result['W2_new'])
    print("Updated b2:\n", result['b2_new'])
    print("Error:", result['error'])
    print("----------------------------------------------------")
    
```
